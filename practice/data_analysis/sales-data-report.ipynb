```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
```

```python
# sales by month
df['month'] = df['order_date'].dt.month

```

```python
df['month'].value_counts().sort_index().plot(kind='bar', title='Sales by Month')
```

```python
# Cell 1: Monthly sales trend and seasonality
monthly = df.groupby('month').agg({'total_price': 'sum', 'quantity': 'sum'}).reset_index()
monthly = monthly.rename(columns={'total_price': 'revenue', 'quantity': 'pizzas_sold'})

monthly.plot(
    x='month',
    y='revenue',
    kind='line',
    marker='o',
    figsize=(12, 5),
    title='Monthly Revenue and Pizza Sales Trend'
)
plt.xticks(rotation=45)
plt.tight_layout()
```

```python
print("\n5. MONTHLY SALES TREND:")
print("   - Sales are generally lower between the 4-5 months.")

print("\n6. SEASONALITY INSIGHTS:")
print("   - Spring (Mar-May): Moderate sales, potential for promotional campaigns.")
print("   - Summer (Jun-Aug): High sales, ideal for seasonal promotions.")
print("   - Fall (Sep-Nov): Steady sales, good for event-based marketing.")
```
