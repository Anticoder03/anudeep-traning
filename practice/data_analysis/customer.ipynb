```python
# cross tab - it is a way to summarize the relationship between two categorical variables. It creates a matrix (or table) that shows the frequency distribution of the variables.
# cross tab - investigate the relationship between Gender
df_crosstab = pd.crosstab(df['Gender'], df['Payment_method'],margins=True,margins_name="Total",normalize=True)*100
```

```python
df_crosstab
```

```python
df_crosstab.plot(kind='bar', figsize=(10, 5), title='Payment Method by Gender',xlabel='Gender',ylabel='Percentage',xlim=(0, 100),ylim=(0, 100))
```

```python
# Another example of cross tab - investigate the relationship between Gender and Employees
```

```python
cross_tab = pd.crosstab(df['Gender'], df['Employees_status'])
cross_tab
```

```python
# ploting chart
cross_tab.plot(kind='bar', figsize=(10, 5), title='Employees Status by Gender')
```
